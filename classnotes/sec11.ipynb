### ACTIVITY #2: Peer Critique

If you are not scheduled for a mentor critique today, then you will participate in a peer critique. Find a partner who is also participating in a peer critique and introduce yourself. The person whose birthday is next will be the first **presenter**, and the other person will be the **reviewer**.

#### How each round works

1. The presenter explains their project goal, what they've built so far, and how to use the project. Then they let the reviewer test and explore it on the presenter's computer. The reviewer can ask clarifying questions during this time. (5-10 min)
2. The reviewer gives feedback using the Ladder of Feedback below. The presenter takes notes in the section below. (5-10 min)
3. Swap roles and repeat.

#### Ladder of Feedback (for the reviewer)

Work through these levels of feedback in order:

1. **Clarifying:** Ask questions to understand the intent of each part of the FP prototype.
    - “What were you trying to accomplish when you…?”
    - “Can you walk me through what happens when…?”
2. **Valuing:** Express what you like about the work using specific details.
    - “I really liked how you… because…”
    - “One thing that stood out to me was…”
3. **Offering concerns:** Raise areas for improvement. Hold off on suggestions.
    - “I wonder if… might cause a problem when…”
    - “I was a little confused by… because…”
4. **Suggesting:** Propose one or two paths forward to address the concerns.
    - “Have you considered trying…?”
    - “One possible next step could be…”

### ACTIVITY #3: Build a Workflow

Open Sid Pardeshi's talk slides, which you can find [here](https://canvas.harvard.edu/courses/173874/files), and turn to slide 16. This slide summarizes how Sid organizes many different AI models in a workflow that allows him to solve everyday problems at little cost. Slides 18-32 use this organizational template to create 7 solutions to some everyday chores.

In this activity, you'll choose a new task and work with AI to create your own **workflow** that'll solve the task. Note that you should **not** build the agents and deterministic tools that would implement the proposed workflow. This means, for your chosen task, that you should:

-   Create an input-team-output diagram like the ones on slides 19, 21, 23, etc.

-   Describe this diagram and the one on slide 16 in a prompt to any AI chatbot. You probably shouldn't use Copilot in your Codespace so that you can save your Codespace minutes and Copilot credits for your final project.

-   Ask the AI to determine and then describe:
    - What the master model will do.
    - How many group leads are needed and what they'll do.
    - How worker models are distributed across these group leads and what they'll do.
    
-   Iterate with the AI until you have a reasonable workflow mapped out that you think will:
    - Effectively solve the task.
    - Do so with little work for the master model.

#### My Chosen Task: Food Card Nutrition Lookup

**Task description:** A user photographs a food card (e.g., a restaurant menu card, a packaged food label, or a cafeteria placard). The system reads the food name from the image, searches a nutrition database for that item, and returns all nutritional information.

---

#### Input–Team–Output Diagram

| WHAT GOES IN | YOUR AGENT TEAM | WHAT COMES OUT |
|:---|:---|:---|
| **Photo of a food card** | **On the laptop:** | **Nutritional profile** |
| A picture taken of a food item's label, menu card, or cafeteria placard | Reads the text on the food card photo. | Food name as read from the card |
| | Cleans and normalizes the extracted food name. | Calories, protein, carbs, fat |
| | Searches the nutrition database for the best matching item. | Vitamins and minerals |
