### Takeaways

Process comments:

1.  It's okay to start with a short prompt.
2.  But don't trust the result without rigorous verification.
3.  Use what you learn from the verification process to determine what information you should have included in your prompt.
4.  Rewrite your prompt and iterate (return to step 2), until the generated program _and_ its output passes your rigorous verification process.

Comments about prompting:

-   Prompting is never a one-and-done process, even if you spend a lot of time in design.
-   Prompting is also not only about how you structure your prompt to get a good AI response.
-   Prompting is part of an iterative process.

### A Failure Taxonomy

Here are five commonly encountered failure modes when **an initial prompt doesn't work**:

1.  Hallucinations and fabricated data

    -  E.g., AI chooses a similarly named key or generates placeholder data.
    -  Live demo? In the directory with `amd_data.json`, prompt AI to "write a python script that gets and compares Apple and AMD revenue data."
    -  Mitigation: Check the code and cross-reference AI outputs against known sources.

2.  Silent failures

    -  The code runs without errors but produces incorrect results. AI gives no assessment of its confidence in the generated program.
    -  Live demo? We saw an example of this with `extract_net_income.py`.
    -  Mitigation: Test with data where you already know the answer.

3.  Misinterpretation of intent

    -  AI understands your words but not your meaning. English is ambiguous. Words like "latest," "current," "recent," and "new" have different meanings in different contexts.
    -  Example: Consider the directive, "Get the latest financial data." It could mean: the most recently filed document, or data from the last calendar year, or the document with the newest "last modified" timestamp.
    -  Mitigation: Ask the AI how it interpreted potentially ambiguous terms and review the AI's code to verify this response.

4.  Hidden assumptions

    -  AI makes important decisions without telling you.  Or it buries these decisions in its many paragraphs of output that you didn’t read!
    -  Example: When you ask for "revenue," AI might assume you want annual revenue (not quarterly), in USD (not other currencies), from operating activities (not total revenue). Only you can determine whether an assumption is reasonable for your use case.
    -  Mitigation: Read the comments and skim through the AI-generated code. Highlight some code and ask AI to explain its assumptions.



5.  Brittle code

    -  The AI-generated code runs and produces correct output for a common example (e.g., the one you included in the prompt), but it breaks on edge cases. AI often optimizes for the example you give it, not for general robustness.
    -  Example: Code that extracts Net Income from AMD's data might fail completely when you try it on a different company because that company uses a different accounting standard, or has a different fiscal year end, or has on file very few years of data, or has years without data.
    -  Mitigation: Test with multiple inputs, including edge cases. Try companies of different sizes, different industries, and different filing histories. We'll talk testing next!

### Revisiting Process

This is where we left the problem-solving process in `cn02`:

1.  Identify a task you want automated
2.  Set the requirements for the tool that'll automate the task
3.  While the tool is incomplete:

    3a.  Pick a(nother) small part and give its requirements to the AI

    3b.  Check what the AI built

4.  Use the tool and focus on the interesting work

It focuses on decomposition and assumes what the AI builds works, which we now know is often true with straightforward problems (e.g., a Mad Libs app) but not with more complex ones.

Step 3b becomes:

-   Recognize when the AI has failed (not always obvious)
-   Diagnose which failure mode occurred
-   Revise your prompt to fix it

This adds a loop within the loop driven by decomposition. It is the more difficult workflow of problem solving with AI.

In section, we'll start on recognition/diagnosing failures with some testing strategies. Soon, we'll introduce a new business problem so you can practice this cycle of identifying a problem, diagnosing it, and eventually overcoming it.
