### Today's Philosophical Musings

We're moving into the realm of agents where capabilities, terminology, and knowledge is changing quickly. Some of you probably have more experience than I do. My goal isn't to be the expert, but to:

-   Make it easy to get started, if you're a novice.
-   Fill some gaps, if you've some experience.

Throughout this, I'll be encouraging you to **think critically** rather than **simply strive for efficiency** (i.e., minimal thinking).

> It's easy but costly to just throw things at AI.

### Example: One-shot the result vs. scaffolded work

Andrew Ng likes to discuss the writing of a research paper and contrast two students' approaches to it.

1.  **One Shot.** One student writes a prompt containing: (1) the teacher's instructions and (2) what he knows from class/life. He ends the prompt directing the AI to write the paper and hopes for a good response.

2.  **Scaffolded Work.** The other student builds a plan for the paper and asks AI to improve it. The plan starts with an AI-assisted research step. Next, with AI help, the student builds a draft, iterates on it, and checks everything along the way. She gets a good response and is sure of it.

Evaluating the two approaches:

-   In (2) , it is easy to check and redirect the AI. Small steps add up to an efficient path to a good paper. Plus, you can answer questions about the final product.
-   In (1), you sink a lot of cost into a single shot at a good first paper. If it is bad, you have little to show for the expense. The large sunk costs might cause you to try and incrementally improve the poor first result. Undoing something bad is harder than carefully building something good.

### How to Train Your Dragon

You've learned about a model's **context** and its **limited context window**. 

-   context = the info a model has when generating a response.
-   context window = the limit on the size of that info.

Your prompt is only part of the context, which includes system prompts, prior messages in this conversation, and the results of tool calls.

----

Let's now add to what we can shove into the model's context. Why? To avoid:

-   Repeating ourselves
-   Forgetting to tell the model something important

In particular, we'll talk about **rules files** and **skills files**.

Key comparisons:

-   Primary purpose
    - Rules: (Should be) unbreakable laws (e.g., guardrails and code styles).
    - Skills: Instructions on how to perform specific actions.
-   Loading mechanism
    - Rules: Always active.
    - Skills: Lazily loaded when a task requires it.
-   Impact on context window
    - Rules: Always there.
    - Skills: Loaded only when triggered.

> **NOTE:** While this is generally true, the specifics vary from vendor to vendor.
>
> Historically, these files have been simple text/markdown files, but this is starting to change. There's also sometimes subtle variation in the directory tree associated with these files.

----

### Rules Files

We won't cover them in any detail because they are **highly personal** or **employer/project specific**.

For example, the next time I teach this material, I'll probably create a global rules files that tells AI to **be concise**.

----

### Skills Files

Skills are like functions. They are:

-   reusable
-   task-specific (i.e., applied when needed)
-   a bundle of instructions (and sometimes including tools/programs)

Two important and distinct pieces to a skill file:

1.  The **description**, which says **when to load the skill**.
2.  The **instructions**, which says **how some work should be done**.

----

### When should you build a skill?

If you find that in a particular circumstance,

1.  You keep repeating the same instructions.
2.  AI defaults to an action, but it isn't what you want.
3.  Everyone on your team needs to get similar results.

### Simple First Skill

> NOTE: I borrowed this example from ["Create your first skill" in the Claude Code Docs](https://code.claude.com/docs/en/skills#create-your-first-skill), which also illustrates the differences I mentioned, i.e., Claude allows an action that isn't supported in GitHub Copilot.

Here's an example. Let's say you want something particular to happen when you're trying to figure out what has changed in your code base since your last commit.

-   Open `m13/.github/skills/summarize-changes/SKILL.md`, which is a skills file placed in a directory tree that GitHub Copilot understands.
-   Top of this file is **YAML frontmatter** (between and including the `---` lines). It must come first!
-   The rest of this file are the instructions, written however you'd like (but they should be easy for an LLM to understand).

----

### YAML Primer

> See [this IBM page](https://www.ibm.com/think/topics/yaml) for a more complete explanation.

The acronym stands for "YAML Ain't Markup Language", and you should think that the language is more like JSON than Markdown.

> Like JSON, YAML is a **data serialization" language, which means that it turns computer data into and out of human-readable text without losing its structure.

Important points about YAML text:

-   Formatting matters! E.g., no tabs, just spaces please.
-   You are looking at name-value pairs.

Important points about the YAML frontmatter in a skill file.

-   Only two names needed:
    - `name`. Its value is 64 characters max. Use only lowercase letters, numbers, and hyphens.
    - `description`. Keep its value to 250 characters for interoperability. The true limit varies between vendors.
-   Notice that we describe **what** the skill does and **when** it should be invoked in the description value.

### Other Considerations

_Do you write these skills files by hand?_

-   No!
-   If you've just gone through a session with an AI that you'd like to turn into a skill, ask the AI to create a skill file from that conversation!

_Skills can live in many places and be distributed in many forms. From Claude Code Docs:_

-   "Save it under your home directory to get it in every project, commit it to a repository to share it with everyone who works there, or distribute it through a plugin or managed settings to reach a whole team." [Link](https://code.claude.com/docs/en/skills#choose-where-skills-load)

_What makes a good skill?_

-   A narrow purpose (e.g., `summarize-changes` is, but `sw-eng-helper` isn't).
-   An explicit procedure that it can perform.
-   Concrete outputs and what to do if there's nothing to output.
-   Keep `SKILL.md` short and focused. Any supporting material goes in other files/folders in the same directory where `SKILL.md` is located.
